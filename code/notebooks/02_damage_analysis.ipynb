```python
# AI Damage Pattern Analysis

This notebook analyzes damage patterns in `dataset/output.csv` for the AI damage claim verification project. We will explore frequency distributions, severity signals, and image quality indicators with both Plotly and Matplotlib.
</VSCode.Cell>
<VSCode.Cell language="python">
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
from pathlib import Path

plt.style.use("dark_background")

DATA_PATH = Path("..") / "dataset" / "output.csv"
df = pd.read_csv(DATA_PATH)

df.head()
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Dataset Snapshot

Review the dataset shape and available columns before diving into damage pattern analysis.
</VSCode.Cell>
<VSCode.Cell language="python">
print("Dataset path:", DATA_PATH)
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

print("\nPreview of top rows:")
df.head()
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Issue Type Frequency

Analyze the most common issue types reported in the dataset. This helps identify dominant damage categories that require the strongest verification workflows.
</VSCode.Cell>
<VSCode.Cell language="python">
if "issue_type" in df.columns:
    issue_counts = df["issue_type"].fillna("Missing").value_counts()
    print(issue_counts)

    fig = px.bar(
        x=issue_counts.index,
        y=issue_counts.values,
        labels={"x": "Issue Type", "y": "Count"},
        title="Issue Type Frequency",
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#fbbf24", text=issue_counts.values, textposition="outside")
    fig.update_layout(xaxis_tickangle=-45, height=520)
    fig.show()

    fig2 = px.pie(
        names=issue_counts.index,
        values=issue_counts.values,
        title="Issue Type Share",
        template="plotly_dark",
    )
    fig2.update_traces(textposition="inside", textinfo="percent+label")
    fig2.show()
else:
    print("Column 'issue_type' not found in dataset.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Severity Level Analysis

Examine the distribution of severity levels to understand how many claims are low, medium, or high severity.
</VSCode.Cell>
<VSCode.Cell language="python">
if "severity" in df.columns:
    severity_counts = df["severity"].fillna("Missing").value_counts()
    print(severity_counts)

    fig = px.bar(
        x=severity_counts.index,
        y=severity_counts.values,
        labels={"x": "Severity", "y": "Count"},
        title="Severity Distribution",
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#ef4444", text=severity_counts.values, textposition="outside")
    fig.update_layout(xaxis_tickangle=-45, height=520)
    fig.show()

    fig3 = px.histogram(
        df,
        x="severity",
        title="Severity Histogram",
        template="plotly_dark",
    )
    fig3.update_layout(xaxis_title="Severity", yaxis_title="Count", height=520)
    fig3.show()
else:
    print("Column 'severity' not found in dataset.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Object Type Categories

Explore the distribution of object types to see which asset categories appear most often in damage claims.
</VSCode.Cell>
<VSCode.Cell language="python">
if "object_type" in df.columns:
    object_counts = df["object_type"].fillna("Missing").value_counts()
    print(object_counts)

    fig = px.bar(
        x=object_counts.index,
        y=object_counts.values,
        labels={"x": "Object Type", "y": "Count"},
        title="Object Type Distribution",
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#38bdf8", text=object_counts.values, textposition="outside")
    fig.update_layout(xaxis_tickangle=-45, height=520)
    fig.show()

    fig2 = px.pie(
        names=object_counts.index,
        values=object_counts.values,
        title="Object Type Share",
        template="plotly_dark",
    )
    fig2.update_traces(textposition="inside", textinfo="percent+label")
    fig2.show()
else:
    print("Column 'object_type' not found in dataset.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Damage Visible Distribution

Evaluate how often damage is detected visually in the claim evidence. This is an important signal for automatic acceptance or manual review.
</VSCode.Cell>
<VSCode.Cell language="python">
if "damage_visible" in df.columns:
    damage_counts = df["damage_visible"].fillna("Missing").astype(str).value_counts()
    print(damage_counts)

    fig = px.bar(
        x=damage_counts.index,
        y=damage_counts.values,
        labels={"x": "Damage Visible", "y": "Count"},
        title="Damage Visible Distribution",
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#a78bfa", text=damage_counts.values, textposition="outside")
    fig.update_layout(xaxis_tickangle=-45, height=520)
    fig.show()

    fig2 = px.pie(
        names=damage_counts.index,
        values=damage_counts.values,
        title="Damage Visible Share",
        template="plotly_dark",
    )
    fig2.update_traces(textposition="inside", textinfo="percent+label")
    fig2.show()
else:
    print("Column 'damage_visible' not found in dataset.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Valid Image Quality

Assess how many images are considered valid. This helps measure data quality and identify cases likely to require manual review.
</VSCode.Cell>
<VSCode.Cell language="python">
if "valid_image" in df.columns:
    valid_counts = df["valid_image"].fillna("Missing").astype(str).value_counts()
    print(valid_counts)

    fig = px.bar(
        x=valid_counts.index,
        y=valid_counts.values,
        labels={"x": "Valid Image", "y": "Count"},
        title="Valid Image Distribution",
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#22c55e", text=valid_counts.values, textposition="outside")
    fig.update_layout(xaxis_tickangle=-45, height=520)
    fig.show()

    fig2 = px.pie(
        names=valid_counts.index,
        values=valid_counts.values,
        title="Valid Image Share",
        template="plotly_dark",
    )
    fig2.update_traces(textposition="inside", textinfo="percent+label")
    fig2.show()

    total_images = len(df)
    valid_images = df["valid_image"].fillna(False).astype(str).isin(["True", "true", "1"]).sum()
    valid_pct = (valid_images / total_images * 100) if total_images else 0
    print(f"\nPercentage of valid images (approx): {valid_pct:.2f}%")
else:
    print("Column 'valid_image' not found in dataset.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Issue vs Severity Heatmap

Visualize the relationship between issue type and severity level with a heatmap to identify high-risk combinations.
</VSCode.Cell>
<VSCode.Cell language="python">
if "issue_type" in df.columns and "severity" in df.columns:
    heatmap_data = (
        df.fillna("Missing").groupby(["issue_type", "severity"]).size().reset_index(name="count")
    )
    pivot = heatmap_data.pivot(index="issue_type", columns="severity", values="count").fillna(0)

    fig = go.Figure(
        data=go.Heatmap(
            z=pivot.values,
            x=pivot.columns.tolist(),
            y=pivot.index.tolist(),
            colorscale="Viridis",
            colorbar=dict(title="Count"),
        )
    )
    fig.update_layout(
        title="Issue Type vs Severity Heatmap",
        xaxis_title="Severity",
        yaxis_title="Issue Type",
        template="plotly_dark",
        height=650,
    )
    fig.show()

    plt.figure(figsize=(12, 8))
    plt.imshow(pivot.values, aspect="auto", cmap="viridis")
    plt.colorbar(label="Count")
    plt.xticks(range(len(pivot.columns)), pivot.columns, rotation=45)
    plt.yticks(range(len(pivot.index)), pivot.index)
    plt.title("Issue Type vs Severity Heatmap")
    plt.tight_layout()
    plt.show()
else:
    print("Required columns for heatmap not found: 'issue_type' and 'severity'.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Manual Review Trends

Inspect manual review trends if a relevant column exists, otherwise infer possible manual review signals from image validity and damage visibility.
</VSCode.Cell>
<VSCode.Cell language="python">
if "manual_review" in df.columns:
    manual_counts = df["manual_review"].fillna("Missing").astype(str).value_counts()
    print(manual_counts)
    fig = px.bar(
        x=manual_counts.index,
        y=manual_counts.values,
        labels={"x": "Manual Review", "y": "Count"},
        title="Manual Review Trends",
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#f97316", text=manual_counts.values, textposition="outside")
    fig.update_layout(xaxis_tickangle=-45, height=520)
    fig.show()
else:
    print("Column 'manual_review' not found. Inferring manual review trends from image quality signals.")
    if "valid_image" in df.columns and "damage_visible" in df.columns:
        inferred = (
            df.assign(
                candidate_manual_review=lambda d: (~d["valid_image"].astype(str).isin(["True", "true", "1"]))
                | (d["damage_visible"].astype(str).isin(["False", "false", "0"]))
            )
        )
        inferred_counts = inferred["candidate_manual_review"].value_counts()
        print(inferred_counts)
        fig = px.bar(
            x=inferred_counts.index.map({True: "Review Likely", False: "Auto Accept Likely"}),
            y=inferred_counts.values,
            labels={"x": "Manual Review Signal", "y": "Count"},
            title="Inferred Manual Review Trend",
            template="plotly_dark",
        )
        fig.update_traces(marker_color=["#f97316", "#22c55e"], text=inferred_counts.values, textposition="outside")
        fig.update_layout(height=520)
        fig.show()
    else:
        print("Unable to infer manual review trends without both 'valid_image' and 'damage_visible'.")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Conclusions

Summarize the most important damage pattern findings and quality signals uncovered in this dataset.
</VSCode.Cell>
<VSCode.Cell language="markdown">
- The most common damage type can be identified from the highest-count `issue_type` category above.
- The most severe damage categories are visible from the severity distribution and issue/severity heatmap, highlighting which issue types are associated with high severity.
- Image quality is critical: the valid image percentage indicates how many claims are supported by usable visual evidence.
- Manual review trends should focus on cases with invalid images or no visible damage, since these are likely to require extra human inspection.
- The combination of issue type and severity is an important risk signal for prioritizing claims and tuning review workflows.
</VSCode.Cell>
```
