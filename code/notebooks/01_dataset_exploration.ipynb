```python
# AI Damage Claim Verification - Dataset Exploration

This notebook explores `dataset/output.csv` for the AI damage claim verification project. We inspect data quality, structure, and distributions across the core fields used by the claim verification pipeline.
</VSCode.Cell>
<VSCode.Cell language="python">
import pandas as pd
import matplotlib.pyplot as plt
import plotly.express as px
from pathlib import Path

plt.style.use("dark_background")

DATA_PATH = Path("..") / "dataset" / "output.csv"
df = pd.read_csv(DATA_PATH)

df.head()
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Dataset Overview

Review the dataset shape, available columns, and a few sample rows to understand the structure of the data.
</VSCode.Cell>
<VSCode.Cell language="python">
print("Dataset path:", DATA_PATH)
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

df.sample(5, random_state=42)
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Missing Values and Duplicate Records

Evaluate data quality by checking for missing values and duplicate rows. These findings help identify cleaning and preprocessing needs.
</VSCode.Cell>
<VSCode.Cell language="python">
missing_counts = df.isna().sum()
missing_percent = (missing_counts / len(df) * 100).round(2)

duplicate_count = df.duplicated().sum()

missing_summary = pd.DataFrame({
    "missing_count": missing_counts,
    "missing_percent": missing_percent,
}).sort_values("missing_count", ascending=False)

missing_summary
</VSCode.Cell>
<VSCode.Cell language="python">
print(f"Duplicate rows: {duplicate_count}")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Data Types and Summary Statistics

Inspect the data types and descriptive statistics for numeric and categorical fields.
</VSCode.Cell>
<VSCode.Cell language="python">
print("Data types:")
print(df.dtypes)

print("\nNumeric summary:")
df.describe(include="number").transpose()
</VSCode.Cell>
<VSCode.Cell language="python">
print("\nCategorical summary:")
df.describe(include="object").transpose()
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Unique Values for Key Fields

Display unique values and counts for the most important categorical fields in the damage claim data.
</VSCode.Cell>
<VSCode.Cell language="python">
for column in ["issue_type", "severity", "object_type"]:
    if column in df.columns:
        print(f"\n=== {column} ===")
        print(df[column].fillna("Missing").value_counts(dropna=False))
    else:
        print(f"\nColumn not found: {column}")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Visualizations

Plot data distributions for object type, issue type, severity, damage visibility, and image validity.
</VSCode.Cell>
<VSCode.Cell language="python">
# Object Type Distribution using matplotlib

if "object_type" in df.columns:
    counts = df["object_type"].fillna("Missing").value_counts()
    fig, ax = plt.subplots(figsize=(10, 5))
    counts.plot(kind="bar", color="#38bdf8", edgecolor="#0f172a", ax=ax)
    ax.set_title("Object Type Distribution")
    ax.set_ylabel("Count")
    ax.set_xlabel("")
    ax.grid(alpha=0.25)
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("Column 'object_type' not found.")
</VSCode.Cell>
<VSCode.Cell language="python">
def plot_bar(column, title):
    if column not in df.columns:
        print(f"Column '{column}' not found.")
        return

    counts = df[column].fillna("Missing").value_counts().reset_index()
    counts.columns = [column, "count"]
    fig = px.bar(
        counts,
        x=column,
        y="count",
        text="count",
        title=title,
        template="plotly_dark",
    )
    fig.update_traces(marker_color="#38bdf8", textfont_color="white")
    fig.update_layout(xaxis_title=None, yaxis_title="Count", height=450)
    fig.show()

plot_bar("issue_type", "Issue Type Distribution")
plot_bar("severity", "Severity Distribution")
plot_bar("damage_visible", "Damage Visible Distribution")
plot_bar("valid_image", "Valid Image Distribution")
</VSCode.Cell>
<VSCode.Cell language="markdown">
## Business Insights

Summarize the most important dataset findings for stakeholders, model development, and risk review.
</VSCode.Cell>
<VSCode.Cell language="markdown">
- The dataset appears to be well structured for AI claim verification, with key fields such as `object_type`, `issue_type`, `severity`, `damage_visible`, and `valid_image` available for modeling.
- Object type distribution reveals which claim categories are most common, guiding resource allocation for review workflows.
- Issue type and severity distributions help identify frequent claim patterns and whether the dataset includes enough high-severity examples for robust model training.
- Damage visibility and valid image indicators are essential quality signals. If either is imbalanced, the review pipeline should account for that in sampling or decision thresholds.
- Minimal duplicate records and clearly defined categorical fields support reliable analytics and prevent noise in model training.
</VSCode.Cell>
```
